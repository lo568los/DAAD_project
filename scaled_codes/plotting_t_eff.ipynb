```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
### Read the dat file and extract the columns
data = np.loadtxt('N6_th0.7853_thk0.5235_teff.txt')

# Extract columns

floquet = data[:, 0]  # first column: floquet

t_eff = data[:, 2]  # last column: t_eff

### Plotting
plt.figure(figsize=(8, 6))
plt.plot(floquet[:20], t_eff[:20], marker='o', linestyle='-', color='b')
plt.axhline(10, color='r', linestyle='--', label=r'$T_K \approx D e^{-1/\rho J}$')  # Add horizontal line at t_eff = 1
plt.xlabel('Floquet Steps')
plt.ylabel('Effective Temperature (t_eff)')
plt.title('Effective Temperature vs Floquet steps')
plt.yscale('log')  # Set y-axis to logarithmic scale
#plt.grid()
plt.show()
```
